```python
import json

with open("data.json", "r", encoding="utf-8") as archivo:
    datos = json.load(archivo)

ventas = datos["ventas"]

print("Número de ventas:", len(ventas))
```

```python
ingresos_productos = {}

for venta in ventas:
    producto = venta["producto"]
    ingresos = venta["cantidad"] * venta["precio_unitario"]

    if producto in ingresos_productos:
        ingresos_productos[producto] = ingresos_productos[producto] + ingresos
    else:
        ingresos_productos[producto] = ingresos

producto_mayor_ingreso = max(ingresos_productos, key=ingresos_productos.get)

print(
    "Producto con más ingresos:",
    producto_mayor_ingreso,
    "-",
    ingresos_productos[producto_mayor_ingreso],
    "€"
)
```

```python
vendedor_buscado = "Ana"

ventas_vendedor = []

for venta in ventas:
    if venta["vendedor"] == vendedor_buscado:
        ventas_vendedor.append(venta)

print("Ventas de", vendedor_buscado, ":")

for venta in ventas_vendedor:
    print(
        venta["producto"],
        "-",
        venta["cantidad"],
        "unidades"
    )
```

```python
total_facturado = 0

for venta in ventas:
    total_facturado = total_facturado + (
        venta["cantidad"] * venta["precio_unitario"]
    )

ticket_medio = total_facturado / len(ventas)

print("Ticket medio:", round(ticket_medio, 2), "€")
```
